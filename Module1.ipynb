```python
import csv
import json

#it is used when the file i have given is invalid
class InvalidFileException(Exception):
    pass
```

```python
#dataloader is used for reading different file formats 
class DataLoader:
    def __init__(self,file_path):
        self.file_path = file_path
        
#to check if the file exists or not 
def validate_file(self):
    try:
        with open(self.file_path,"r"):
            return True
    except FileNotFoundError:
        raise InvalidFileException("file not found")

```

```python
#to read csv file
def read_csv(self):
    self.validate_file()
    with open(self.file_path,"r") as file:
        return list(csv.reader(file))
```

```python
# Read JSON file
def read_json(self):
    self.validate_file()

    with open(self.file_path, "r") as file:
        return json.load(file)
```

```python
 # Read TXT file
def read_txt(self):
    self.validate_file()

    with open(self.file_path, "r") as file:
        return file.read()
```

```python
# Load the file based on its format
def load(self):

    if self.file_path.endswith(".csv"):
         return self.read_csv()

    elif self.file_path.endswith(".json"):
        return self.read_json()

    elif self.file_path.endswith(".txt"):
        return self.read_txt()

    else:
        raise InvalidFileException("Unsupported file format")
```
