validate_file() is a method used to check whether the given file exists or not.
try is used to test a block of code that may cause an error.
except is used to handle errors that occur inside the try block.
raise is used to manually raise an exception.

```python

        
#to check if the file exists or not 
def validate_file(self):
    try:
        with open(self.file_path,"r"):
            return True
    except FileNotFoundError:
        raise InvalidFileException("file not found")

```

what is csv?comma seperated values used to store data in rows and columns.
CSV is a module used to read and write data in CSV files.
how to read csv file?
Checks the file, opens it in read mode, reads the CSV data and returns it as a list.

```python
#to read csv file
def read_csv(self):
    self.validate_file()
    with open(self.file_path,"r") as file:
        return list(csv.reader(file))
```

read_json() is a method used to read data from a JSON file.
It first checks whether the file exists and then opens the file in read mode.
json.load() is used to read the JSON data and return it to the program.

```python
# Read JSON file
def read_json(self):
    self.validate_file()

    with open(self.file_path, "r") as file:
        return json.load(file)
```

read_txt() is a method used to read data from a TXT file.
It first checks whether the file exists and then opens the file in read mode.
file.read() is used to read the complete text from the file and return it to the program.

```python
 # Read TXT file
def read_txt(self):
    self.validate_file()

    with open(self.file_path, "r") as file:
        return file.read()
```

load() is a method used to identify the file format and call the appropriate reading method.
It checks whether the file is CSV JSON or TXT using endswith().
If the format is not supported it raises InvalidFileException.

```python
# Load the file based on its format
def load(self):

    if self.file_path.endswith(".csv"):
         return self.read_csv()

    elif self.file_path.endswith(".json"):
        return self.read_json()

    elif self.file_path.endswith(".txt"):
        return self.read_txt()

    else:
        raise InvalidFileException("Unsupported file format")
```
