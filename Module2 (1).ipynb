What is data writer?
DataWriter is used to save data into different file formats and it can save data as CSV file,Json file,TXT file

```python
import csv
import json
import pickle
```

csv → Used to work with CSV files.
json → JSON stands for JavaScript Object Notation and is used to work with JSON data.
pickle → Used to store Python objects in a file.

```python
class DataWriter:
    pass
```

save_csv() is a method used to save data into a CSV file.
It opens the given file in write mode and uses csv.writer() to write the data.
writerows() is used to write multiple rows into the CSV file.

```python
def save_csv(self, data, file_path):
    with open(file_path, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerows(data)
```

save_json() is a method used to save data into a JSON file.
It uses json.dump() to write Python data into JSON format.
The indent=4 parameter makes the JSON file properly formatted and readable.

```python
def save_json(self, data, file_path):
    with open(file_path, "w") as file:
        json.dump(data, file, indent=4)
```

save_pickle() is a method used to save Python objects into a Pickle file.
It uses pickle.dump() to serialize and store the Python object.
"wb" means write binary mode, which is required for Pickle files.

```python
def save_pickle(self, data, file_path):
    with open(file_path, "wb") as file:
        pickle.dump(data, file)
```

save_txt() is a method used to save data into a text file.
It uses file.write() to write the given data into the file.
"w" means write mode, which allows the file to be created or overwritten.

```python
def save_txt(self, data, file_path):
    with open(file_path, "w") as file:
        file.write(data)
```
